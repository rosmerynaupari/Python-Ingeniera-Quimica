## A2. Ejemplo de ingeniería química: concentración (2 min)

Para una solución con 15 g de NaCl en un volumen **final** de 250 mL:

\[C\,(g/L)=rac{m\,(g)}{V\,(L)}\]

Convierte primero los mililitros a litros.

```python
masa_nacl_g = 15
volumen_solucion_ml = 250
volumen_solucion_l = volumen_solucion_ml / 1000
concentracion_g_l = masa_nacl_g / volumen_solucion_l
print("Concentración:", concentracion_g_l, "g/L")  # 60.0 g/L
```

## A3. Formulación de biopelículas: porcentaje de glicerol (2 min)

En este taller el porcentaje de glicerol se define **respecto a la masa seca de almidón** (no respecto a la masa total de la mezcla):

\[\%G=rac{m_{glicerol}}{m_{almidón\ seco}}	imes 100\]

```python
masa_almidon_basico_g = 5.0
masa_glicerol_basico_g = 1.0
porcentaje_glicerol_basico = masa_glicerol_basico_g / masa_almidon_basico_g * 100
print("Glicerol respecto al almidón seco:", porcentaje_glicerol_basico, "%")
```

## A4. Comparaciones y decisiones: `if` / `else` (2 min)

Operadores: `>`, `<`, `>=`, `<=`, `==` (igualdad), `!=` (diferente). Las comparaciones devuelven `True` o `False`.

**Nota:** el límite de 20 MPa se usa solo como criterio didáctico; no es un requisito universal para biopelículas.

```python
resistencia_mpa = 24
limite_didactico_mpa = 20
print("¿Cumple?:", resistencia_mpa >= limite_didactico_mpa)
print("¿Son iguales?:", resistencia_mpa == limite_didactico_mpa)

if resistencia_mpa >= limite_didactico_mpa:
    print("La biopelícula cumple el criterio didáctico")
else:
    print("La biopelícula no cumple el criterio didáctico")
```

Una **lista** almacena varios valores. Un bucle `for` repite el mismo cálculo para cada elemento.

## A6. Crear funciones reutilizables (3 min)

```python
def calcular_glicerol(masa_almidon_g, porcentaje):
    """Devuelve la masa de glicerol en gramos."""
    return masa_almidon_g * porcentaje / 100

print("Formulación 1:", calcular_glicerol(5, 20), "g")
print("Formulación 2:", calcular_glicerol(10, 25), "g")
```

## 4. Modelamiento: caso didáctico de biopelículas (15 min)
**Variables de entrada:** glicerol `G` (% sobre almidón seco), temperatura de secado `T` (°C) y tiempo `t` (h).

**Salidas:** resistencia a tracción `σ` (MPa) y elongación a rotura `ε` (%).

**Rangos de exploración didácticos:** `10 ≤ G ≤ 30`, `30 ≤ T ≤ 60`, `12 ≤ t ≤ 24`.

Normalización: `x1=(G−20)/10`, `x2=(T−45)/15`, `x3=(t−18)/6`.

Modelos polinomiales hipotéticos:

- `σ = 24 − 6x1 + 2x2 + x3 − 2x1² − x2² − 0.5x3²`
- `ε = 30 + 12x1 − 2x2 − x3 − 3x1² − 2x2² − x3²`

**Estos coeficientes son inventados para practicar programación; no son correlaciones validadas.**


```python
def propiedades(G, T, t):
    x1 = (G - 20) / 10
    x2 = (T - 45) / 15
    x3 = (t - 18) / 6
    resistencia = 24 - 6*x1 + 2*x2 + x3 - 2*x1**2 - x2**2 - 0.5*x3**2
    elongacion = 30 + 12*x1 - 2*x2 - x3 - 3*x1**2 - 2*x2**2 - x3**2
    return resistencia, elongacion

sigma, epsilon = propiedades(20, 45, 18)
print(f'Resistencia: {sigma:.2f} MPa; elongación: {epsilon:.2f} %')
```

## 5. Matplotlib: representar resultados (10 min)
Visualizamos por separado las dos propiedades, porque tienen unidades distintas.


```python
G = np.linspace(10, 30, 100)
resistencia, elongacion = propiedades(G, 45, 18)
fig, ejes = plt.subplots(1, 2, figsize=(12, 4))
ejes[0].plot(G, resistencia)
ejes[0].set(xlabel='Glicerol (% sobre almidón seco)', ylabel='Resistencia (MPa)', title='Resistencia vs. glicerol')
ejes[1].plot(G, elongacion)
ejes[1].set(xlabel='Glicerol (% sobre almidón seco)', ylabel='Elongación (%)', title='Elongación vs. glicerol')
for ax in ejes: ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

## 6. Simulación con NumPy y Pandas (10 min)
Generamos **2000 combinaciones virtuales**, no 2000 experimentos.


```python
rng = np.random.default_rng(42)
n = 2000
G_sim = rng.uniform(10, 30, n)
T_sim = rng.uniform(30, 60, n)
t_sim = rng.uniform(12, 24, n)
sigma_sim, epsilon_sim = propiedades(G_sim, T_sim, t_sim)
datos = pd.DataFrame({'Glicerol_pct': G_sim, 'Temperatura_C': T_sim, 'Tiempo_h': t_sim, 'Resistencia_MPa': sigma_sim, 'Elongacion_pct': epsilon_sim})
display(datos.head())
print('Filas simuladas:', len(datos))
```

## 7. Seaborn: relaciones entre variables (5 min)
La matriz de correlaciones resume asociaciones **dentro del modelo didáctico**.


```python
plt.figure(figsize=(8, 5))
sns.heatmap(datos.corr(numeric_only=True), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('Correlaciones de las variables simuladas')
plt.tight_layout()
plt.show()
```

## 8. SciPy: optimización con restricción (15 min)
**Objetivo:** maximizar elongación, manteniendo una resistencia de al menos **20 MPa**. Este umbral se elige únicamente para el ejercicio.

Usaremos `differential_evolution` con una **restricción explícita**, para no depender solo de una penalización.


```python
from scipy.optimize import NonlinearConstraint

def objetivo(x):
    _, elongacion = propiedades(*x)
    return -float(elongacion)

def resistencia_restringida(x):
    resistencia, _ = propiedades(*x)
    return float(resistencia)

restriccion = NonlinearConstraint(resistencia_restringida, 20, np.inf)
limites = [(10, 30), (30, 60), (12, 24)]
resultado = differential_evolution(objetivo, limites, constraints=(restriccion,), seed=42, tol=1e-8)
G_opt, T_opt, t_opt = resultado.x
sigma_opt, epsilon_opt = propiedades(*resultado.x)
print('Convergencia:', resultado.success)
print(f'Glicerol: {G_opt:.2f} %; T: {T_opt:.2f} °C; t: {t_opt:.2f} h')
print(f'Resistencia: {sigma_opt:.2f} MPa; elongación: {epsilon_opt:.2f} %')
print('¿Cumple la restricción?', sigma_opt >= 20 - 1e-6)
```

## 9. SymPy: una derivada simbólica (5 min)
Las derivadas permiten analizar cómo cambia una respuesta respecto de una variable.


```python
g = sp.symbols('g', real=True)
x = (g - 20)/10
elongacion_simbolica = 30 + 12*x - 3*x**2  # T=45 °C, t=18 h
derivada = sp.diff(elongacion_simbolica, g)
print('d(elongación)/d(glicerol) =', derivada)
print('En G=20 %:', derivada.subs(g, 20))
```
